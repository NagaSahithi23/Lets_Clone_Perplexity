```python
#PROGRAM THAT PRINTS REVERSE OF A NUMBER USING FUNCTIONS
def reverse(a):
  rev=0
  while a!=0:
    rem = a % 10
    rev = rev*10+ rem
    a = a//10
  print(rev)
reverse(234)
```
