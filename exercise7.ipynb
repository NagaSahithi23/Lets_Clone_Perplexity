```python
#OPERATIONS ON SETS
a1={1,2,3,4,5,"girl"}
a2={44,66,77,5,88,"boy"}
print(a1|a2)
print(a1&a2)
print(a1-a2)
print(a1^a2)
```

```python
#METHODS
s1={1,2,3,4,5}
s2={5,6,7,8,9}
print(s1.union(s2))
print(s1.intersection(s2))
print(s1.difference(s2))
print(s1.symmetric_difference(s2))
print(s1.remove(1))
print(s1.discard(2))
print(s1.pop())
print(s1.clear())

```
